# Defoliation Timing
Both spongy moth egg hatch and host tree leaf out vary spatially in response to temperature variation. Thus, timing of defoliation may also vary spatially. Understanding this variation is critical for maximizing estimates of defoliation from remote sensing.

```python
sites = [allegheny, 
         fingerlakes1, 
         fingerlakes2, 
         hudsonvalley1, 
         hudsonvalley2, 
         lakegeorge, 
         michigan1, 
         michigan2, 
         ontario1, 
         ontario2, 
         pennsylvania,
         virginia, 
         wisconsin]

sites_ordered = [virginia,
         pennsylvania,
         hudsonvalley1, 
         hudsonvalley2, 
         allegheny,
         fingerlakes2,
         fingerlakes1,
         michigan1,
         lakegeorge, 
         michigan2, 
         ontario1, 
         ontario2,
         wisconsin]
```

```python
def calculate_defoliation(sites, windows):
    defols = np.zeros((len(sites), len(windows)))
    for i, window in enumerate(windows):
        for j, site in enumerate(sites):
            defol = site.loc[(site['doy'] >= window[0]) & (site['doy'] <= window[1]), 'EVI'].mean()
            defols[j, i] = defol
    return defols
```

```python
def rank_rows(values):
    return np.argsort(np.argsort(values, axis=1), axis=1)
```

## Example 1 month window selection

```python
vals = calculate_defoliation(sites_ordered, 
                             [(150, 180), (160, 190), (170, 200), 
                              (180, 210), (190, 220)])
ranks = rank_rows(vals)
```

```python
fig, ax = uplt.subplots()
m = ax.imshow(vals, cmap='reds_r')
mask = np.isnan(vals)
mask2 = mask.copy().astype(np.float64)
mask2[~mask] = np.nan
ax.imshow(mask2, color='blue')
ax.colorbar(m, label='EVI')

ax.format(xticklabels=['150-180', '160-190', '170-200', '180-210', '190-220'],
          xticks=[0, 1, 2, 3, 4], xlabel='Window (DOY)',
          yticklabels=['Virginia', 'Pennsylvania', 'Hudson Valley 1', 'Hudson Valley 2', 'Allegheny',
                      'Finger Lakes 2', 'Finger Lakes 1', 'Michigan 1', 'Lake George', 'Michigan 2', 
                      'Ontario 1', 'Ontario 2', 'Wisconsin'],
          yticks=np.arange(len(sites_ordered)), ylabel='Site',
          suptitle='Mean EVI in Different 30-Day Windows')
```

```python
fig, ax = uplt.subplots()
m = ax.imshow(ranks, cmap='reds')
ax.imshow(mask2, color='blue')
ax.colorbar(m, label='Rank', ticks=[0.4, 1.2, 2.0, 2.8, 3.6], ticklabels=['1', '2', '3', '4', '5'])

ax.format(xticklabels=['150-180', '160-190', '170-200', '180-210', '190-220'],
          xticks=[0, 1, 2, 3, 4], xlabel='Window (DOY)',
          yticklabels=['Virginia', 'Pennsylvania', 'Hudson Valley 1', 'Hudson Valley 2', 'Allegheny',
                      'Finger Lakes 2', 'Finger Lakes 1', 'Michigan 1', 'Lake George', 'Michigan 2', 
                      'Ontario 1', 'Ontario 2', 'Wisconsin'],
          yticks=np.arange(len(sites_ordered)), ylabel='Site',
          suptitle='Mean EVI Ranks in Different 30-Day Windows')
```
