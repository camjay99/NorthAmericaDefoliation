```python
%load_ext autoreload
%autoreload 2

import pickle

import geopandas as gpd
import pandas as pd
import ultraplot as uplt
i

import utils.sif_processing as sif
```

```python
regions = [('Adirondacks1_2021', 2021),
           ('Adirondacks2_2021', 2022),
           ('Allegheny_Reservoir_2021', 2021),
           ('Finger_Lakes_2021', 2021),
           ('Hudson_Valley_2023', 2023),
           ('Hudson_Valley_2024', 2024),
           ('Lake_Champlain_2022', 2022),
           ('Lake_George_2022', 2022),
           ('Michigan2_2021', 2021),
           ('Michigan2_2022', 2022),
           ('Michigan3_2020', 2020),
           ('Michigan3_2021', 2021),
           ('New_Hampshire_2022', 2022),
           ('New_Hampshire_Maine_2021', 2021),
           ('Pennsylvania2_2023', 2023),
           ('Southern_Tier_2022', 2022),
           ('Virginia_2024', 2024),
           ('Wisconsin_Michigan_2022', 2022),
           ('Wisconsin1_2022', 2022),]
mask = '0_10'
```

```python
decline = {}
for region, year in regions:
    with open(f'../Data/SIF/{mask}_mask/SingleObs_Poly_TROPOMISIF743daily_743ErrorSmaller10_SIFin-2to5_{region}_{mask}_US_2019_2025_WTOA_new.pkl', 'rb') as f:
        data = pickle.load(f)
    sif_df = data[['SIF_743', 'SIF_ERROR_743', 'Mean_TOA_RAD_743','Day', 'Month', 'Year', 'geometry', 'Overlap_Area_cea']].copy()
    sif_df['SIF_743_TOAnorm'] = sif_df['SIF_743'] / sif_df['Mean_TOA_RAD_743']
    sif_df['date'] = pd.to_datetime(sif_df[['Year', 'Month', 'Day']])
    sif_df = sif_df.set_index('date')
    print(len(sif_df), region, year)

    decline[region] = sif.get_sif_anomalies(sif_df, ['SIF_743_TOAnorm'], weights='Overlap_Area_cea', error=None)[year]
```

```python
decline_adj = {}
for region, year in regions:
    with open(f'../Data/SIF/{mask}_mask/SingleObs_Poly_TROPOMISIF743daily_743ErrorSmaller10_SIFin-2to5_{region}_{mask}_US_2019_2025_WTOA_new.pkl', 'rb') as f:
        data = pickle.load(f)
    sif_df = data[['SIF_743', 'SIF_ERROR_743', 'Mean_TOA_RAD_743','Day', 'Month', 'Year', 'geometry', 'Overlap_Area_cea']].copy()
    sif_df = sif_df.to_crs(epsg=5070)
    sif_df['total_area'] = sif_df.geometry.area
    sif_df['adjust_factor'] = (sif_df['Overlap_Area_cea']/sif_df['total_area'])
    sif_df['SIF_743_TOAnorm'] = sif_df['SIF_743_adjusted'] / sif_df['Mean_TOA_RAD_743']
    sif_df['date'] = pd.to_datetime(sif_df[['Year', 'Month', 'Day']])
    sif_df = sif_df.set_index('date')
    print(len(sif_df), region, year)
    print(sif_df.head())

    decline_adj[region] = sif.get_sif_anomalies(sif_df, ['SIF_743_TOAnorm'], weights='Overlap_Area_cea', error=None)[year]
```
